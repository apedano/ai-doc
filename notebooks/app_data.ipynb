# Data about app stores

## Import
Read csv in panda

```python
import pandas as pd

# Show numeric output in decimal format e.g., 2.15
pd.options.display.float_format = '{:,.2f}'.format
df_apps = pd.read_csv('../dive-into-deep-learning/docs/002-preliminaries/apps.csv')
df_apps.head()
```

### Drop Unused Columns

**Challenge**: Remove the columns called `Last_Updated` and `Android_Version` from the DataFrame. We will not use these columns.

```python
df_apps.drop(columns=["Last_Updated", "Android_Ver"], inplace=True)
```

### Find and Remove NaN values in Ratings

**Challenge**: How may rows have a NaN value (not-a-number) in the Ratings column? Create DataFrame called `df_apps_clean` that does not include these rows.


```python
df_apps.Rating.isna().value_counts()
```

```python
df_apps = df_apps[df_apps['Rating'].isna() == False]
df_apps.Rating.isna().value_counts()
```

## Remove duplicate rows

```python
df_apps.drop_duplicates(inplace=True)
```
