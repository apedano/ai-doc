```python
import numpy as np
import matplotlib.pyplot as plt
def plotSinusoid(k, d=512, n=10000):
    x = np.arange(0, 512, 1)
    denominator = np.power(n, 2*x/d)
    y = np.sin(k/denominator)
    plt.plot(x, y)
    plt.title('k = ' + str(k))

fig = plt.figure(figsize=(15, 4))
for i in range(4):
    plt.subplot(141 + i)
    plotSinusoid(i*4)
```
